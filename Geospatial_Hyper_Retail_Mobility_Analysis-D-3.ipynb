```python
df['utcTimestamp'] = pd.to_datetime(df['utcTimestamp'])
```

```python
df['local_time'] = df['utcTimestamp']+pd.to_timedelta(df['timezoneOffset'],unit='m')
```

```python
df['utcTimestamp'].head(10)
```

```python
df['local_time'].head(10)
```

```python
df['hour'] = df['local_time'].dt.hour
df['hour'].head(10)
```

```python
df['is_week'] = df['local_time'].dt.day_of_week.isin([5,6]).astype(int)
df['is_week'].head(10)
```

```python
df['day_of_week'] = df['local_time'].dt.day_name()
df['day_of_week'].head(10)
```

```python
coffee_shop = df[df['venueCategory'].str.contains('Coffee Shop',case=False,na = False)].copy()
coffee_shop.head(10)
```

```python
print(f"Total NYC Check-Ins: {len(df)}")
print(f"Total Coffee Shop Check-Ins: {len(coffee_shop)}")
print(f"Unique Coffee Shop Locations: {coffee_shop['venueId'].nunique()}")
```

```python
coffee_shop_metrics = coffee_shop.groupby(['venueId','latitude','longitude'])
coffee_shop_metrics.head(10)
```

```python
coffee_shop.shape
```

```python
shop_agg = coffee_shop_metrics.agg(total_footfall=('userId', 'count'),unique_visitors=('userId', 'nunique'),peak_hours=('hour', lambda x: x.mode()[0] if not x.mode().empty else None),weekend_share=('is_week', 'mean')).reset_index()
shop_agg.head(10)
```

```python
shop_agg['repeat_ratio'] = (shop_agg['total_footfall'] / shop_agg['unique_visitors']).round(2)
shop_agg[['venueId', 'total_footfall', 'unique_visitors', 'repeat_ratio']].head(10)
```

```python
shop_agg['performance_tier'] = pd.qcut(shop_agg['total_footfall'],q=3,labels=['Low footfall (Risk)','Moderate footfall','High footfall (Target)'])
print(shop_agg.sort_values('total_footfall',ascending=False).head(10))
```
