```python
df['utcTimestamp'] = pd.to_datetime(df['utcTimestamp'])
```

```python
df['local_time'] = df['utcTimestamp']+pd.to_timedelta(df['timezoneOffset'],unit='m')
```

```python
df['hour'] = df['local_time'].dt.hour
df['hour'].head(10)
```

```python
df['is_week'] = df['local_time'].dt.day_of_week.isin([5,6]).astype(int)
df['is_week'].head(10)
```

```python
df['day_of_week'] = df['local_time'].dt.day_name()
df['day_of_week'].head(10)
```

```python
coffee_shop = df[df['venueCategory'].str.contains('Coffee Shop',case=False,na = False)].copy()
coffee_shop.head(10)
```

```python
print(f"Total NYC Check-Ins: {len(df)}")
print(f"Total Coffee Shop Check-Ins: {len(coffee_shop)}")
print(f"Unique Coffee Shop Locations: {coffee_shop['venueId'].nunique()}")
```

```python
coffee_shop_metrics = coffee_shop.groupby(['venueId','latitude','longitude'])
coffee_shop_metrics.head(10)
```

```python
coffee_shop.shape
```

```python
shop_agg = coffee_shop_metrics.agg(total_footfall=('userId', 'count'),unique_visitors=('userId', 'nunique'),peak_hours=('hour', lambda x: x.mode()[0] if not x.mode().empty else None),weekend_share=('is_week', 'mean')).reset_index()
shop_agg.head(10)
```

```python
shop_agg['repeat_ratio'] = (shop_agg['total_footfall'] / shop_agg['unique_visitors']).round(2)
shop_agg[['venueId', 'total_footfall', 'unique_visitors', 'repeat_ratio']].head(10)
```

```python
shop_agg['performance_tier'] = pd.qcut(shop_agg['total_footfall'],q=3,labels=['Low footfall (Risk)','Moderate footfall','High footfall (Target)'])
print(shop_agg.sort_values('total_footfall',ascending=False).head(10))
```

```python
from sklearn.preprocessing import StandardScaler
numeric_features = ['total_footfall', 'repeat_ratio']
SS = StandardScaler()
shop_agg[[f'{col}_scaled' for col in numeric_features]] = SS.fit_transform(shop_agg[numeric_features])
shop_agg.head(10)
```

```python
from sklearn.cluster import KMeans
kmeans = KMeans(n_clusters=3,random_state=42)
shop_agg['cluster'] = kmeans.fit_predict(shop_agg[numeric_features])
print(shop_agg)
```

```python
shop_agg = coffee_shop.groupby('venueId').agg(latitude=('latitude','mean'),longitude=('longitude','mean'),total_footfall=('userId','count'),unique_visitors=('userId','nunique'),peak_hour=('hour',lambda x: x.mode()[0] if not x.mode().empty else 12),weekend_share=('is_week','mean')).reset_index()
```

```python
shop_agg['repeat_ratio'] = (shop_agg['total_footfall']/shop_agg['unique_visitors']).round(2)
```

```python
features = ['latitude','longitude','total_footfall','repeat_ratio','weekend_share','peak_hour']
SS = StandardScaler()
X = SS.fit_transform(shop_agg[features])
```

```python
import matplotlib.pyplot as plt
wcss = []
range = range(2,10)
for k in range:
  km = KMeans(n_clusters=k, random_state=42, n_init=10)
  km.fit(X)
  wcss.append(km.inertia_)
plt.figure(figsize=(7,4))
plt.plot(range, wcss, marker='o')
plt.title('Elbow method for Clusters')
plt.xlabel('Number of Clusters')
plt.ylabel('WCSS')
plt.show()
```
